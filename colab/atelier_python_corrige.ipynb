## Partie 1 : Syntaxe de base, variables et structures de contrôle (Bloc 3)

Découvrons comment Python gère les variables, les types de données, les calculs, les conditions et les boucles.

### 1.1 Variables, types de base et typage dynamique mais FORT

* Pas de déclaration de type préalable (`int x = 5` en C#/Java devient simplement `x = 5`).
* Convention officielle : les variables s'écrivent en **`snake_case`**.
* Le typage est **dynamique** (le type est lié à la valeur en mémoire) mais **fort** (Python refuse d'additionner implicitement du texte et un nombre, contrairement à JavaScript).

```python
# --- 1.1 Variables et types fondamentaux ---

nom_serveur = "prod-mtl-web01"   # str (chaîne de caractères)
port_ecoute = 8443               # int (entier)
charge_cpu = 68.459              # float (nombre à virgule flottante)
est_en_ligne = True              # bool (True ou False, avec majuscule !)
certificat_ssl = None            # NoneType (l'équivalent de null)

# Affichage des valeurs et de leurs types réels détectés par Python
print("Serveur      :", nom_serveur, "-> Type :", type(nom_serveur))
print("Port d'écoute:", port_ecoute, "-> Type :", type(port_ecoute))
print("Charge CPU   :", charge_cpu, "-> Type :", type(charge_cpu))
print("En ligne ?   :", est_en_ligne, "-> Type :", type(est_en_ligne))
print("Certificat   :", certificat_ssl, "-> Type :", type(certificat_ssl))
```

#### Démonstration du typage fort : conversion explicite
En JavaScript, `"150" + 25` donne `"15025"`. En Python, cela génère une erreur `TypeError`. Il faut convertir explicitement avec `int()` :

```python
# --- Démonstration du typage fort ---

nb_requetes_recues = "150"    # Reçu sous forme de chaîne de caractères (str)
nouvelles_requetes = 25       # Entier (int)

# Conversion explicite avec int() pour permettre l'addition mathématique
total = int(nb_requetes_recues) + nouvelles_requetes
print("Total des requêtes calculé :", total)

# 💡 À essayer : Décommentez la ligne suivante pour voir comment Python bloque l'addition non sécuritaire :
# print(nb_requetes_recues + nouvelles_requetes)
```

### 1.2 Opérateurs mathématiques et chaînes formatées (*f-strings*)

* **Division réelle `/`** : Donne **toujours** un nombre décimal (`float`), même si les deux nombres sont des entiers (ex. : `17 / 5` = `3.4`).
* **Division entière `//`** : Donne le quotient entier (`17 // 5` = `3`).
* **Puissance `**`** : `2 ** 10` calcule 2 exposant 10 (= 1024).
* **Les *f-strings* (`f"..."`)** : Permettent d'insérer des expressions directement entre accolades `{}` avec des formateurs comme `:.2f` (2 chiffres après la virgule).

```python
# --- 1.2 Calculs et formatage f-string ---

espace_total_go = 500
espace_utilise_go = 373

# Calculs mathématiques
pourcentage_utilise = (espace_utilise_go / espace_total_go) * 100
division_entiere = 17 // 5
reste_modulo = 17 % 5
capacite_adresses_ipv4 = 2 ** 32  # 2 exposant 32

print(f"Espace disque utilisé : {pourcentage_utilise:.1f} %")
print(f"Division entière 17 // 5 = {division_entiere} (reste : {reste_modulo})")
print(f"Nombre total d'adresses IPv4 : {capacite_adresses_ipv4:,} adresses")

# 💡 À essayer : Modifiez 'espace_utilise_go = 450' et réexécutez la cellule pour voir le nouveau pourcentage !
```

### 1.3 Prise de décision : `if`, `elif`, `else` et comparaison chaînée

* Pas d'accolades : le bloc d'instructions est **indenté de 4 espaces** après le deux-points `:`.
* Opérateurs logiques écrits en lettres : `and`, `or`, `not`.
* **Comparaison chaînée** élégante : on peut écrire `200 <= code < 300` directement !
* **Truthiness** : En Python, une collection vide (`[]`), une chaîne vide (`""`), le nombre `0` et `None` sont automatiquement évalués à `False`.

```python
# --- 1.3 Conditions et décisions ---

code_statut_http = 404
est_service_critique = True

if 200 <= code_statut_http < 300:
    etat = "Succès"
    niveau_gravite = "NORMAL"
elif code_statut_http == 404:
    etat = "Ressource introuvable"
    niveau_gravite = "AVERTISSEMENT"
elif code_statut_http >= 500:
    etat = "Erreur interne du serveur"
    niveau_gravite = "CRITIQUE"
else:
    etat = "Autre statut"
    niveau_gravite = "INFO"

# Utilisation de 'and' en toutes lettres et opérateur ternaire en une ligne
notifier_equipe = (niveau_gravite == "CRITIQUE") or (niveau_gravite == "AVERTISSEMENT" and est_service_critique)
action_requise = "Envoyer alerte par courriel" if notifier_equipe else "Aucune alerte"

print(f"Statut HTTP {code_statut_http} : {etat}")
print(f"Gravité : [{niveau_gravite}] -> Action : {action_requise}")

# 💡 À essayer : Remplacez 'code_statut_http = 503' ou '200' et observez le changement d'action !
```

### 1.4 Les boucles : l'itération à la façon de Python

* En Python, la boucle `for` est toujours un **foreach** : elle parcourt directement les éléments d'une collection.
* Si on a besoin d'un compteur numérique : on utilise `range(debut, fin_exclue)`.
* Si on a besoin de **l'indice ET de la valeur** : on utilise `enumerate(collection, start=1)`.

```python
# --- 1.4 Boucles d'itération ---

noeuds_cluster = ["noeud-mtl-01", "noeud-mtl-02", "noeud-qc-01", "noeud-gatineau-01"]

print("--- 1. Parcours direct des éléments (foreach) ---")
for noeud in noeuds_cluster:
    print(f"Ping vers le serveur {noeud}...")

print("\n--- 2. Parcours avec indice et valeur grâce à enumerate() ---")
for index, noeud in enumerate(noeuds_cluster, start=1):
    print(f"Serveur #{index} : {noeud}")

print("\n--- 3. Répéter une action N fois avec range() ---")
for essai in range(1, 4):
    print(f"Tentative de connexion #{essai}/3")
```

### 1.5 Exemple intégrateur 1 : Analyse automatique de journaux serveur (*logs*)

Voici un script complet qui analyse une liste d'événements texte formatés comme ceci : `"NIVEAU|HORODATAGE|MESSAGE|LATENCE_MS"`.
Exécutez la cellule pour observer comment Python découpe les chaînes avec `.split("|")`, déballe les variables, compte les erreurs et extrait la latence maximale.

```python
# --- 1.5 Analyseur de journaux d'événements serveur ---

journaux_bruts = [
    "INFO|2026-10-16 08:15:02|Authentification réussie pour 'admin'|18",
    "WARNING|2026-10-16 08:16:40|Mémoire cache Redis à 84 %|45",
    "ERROR|2026-10-16 08:17:10|Connexion interrompue à PostgreSQL|1420",
    "INFO|2026-10-16 08:18:05|Requête API /v1/factures traitée|32",
    "ERROR|2026-10-16 08:19:55|Délai dépassé passerelle de paiement|2500",
    "INFO|2026-10-16 08:20:12|Fermeture de session utilisateur|11",
]

total_evenements = len(journaux_bruts)
nombre_erreurs = 0
latence_maximale = 0
somme_latences = 0

print("Détail des incidents détectés :")
print("-" * 60)

for entree in journaux_bruts:
    # Déballage direct des 4 champs découpés par le séparateur '|'
    niveau, date_heure, message, latence_str = entree.split("|")
    latence_ms = int(latence_str)
    somme_latences += latence_ms

    if niveau == "ERROR":
        nombre_erreurs += 1
        print(f"🚨 [ERREUR] {date_heure} : {message} ({latence_ms} ms)")
    
    if latence_ms > latence_maximale:
        latence_maximale = latence_ms

latence_moyenne = somme_latences / total_evenements

print("-" * 60)
print(f"Bilan : {nombre_erreurs} erreur(s) détectée(s) sur {total_evenements} événements analysés.")
print(f"Latence moyenne : {latence_moyenne:.1f} ms | Latence maximale : {latence_maximale} ms")
```

## Partie 2 : Structures de données, fonctions et code « pythonique » (Bloc 4)

Explorons les 4 collections intégrées de Python (`list`, `dict`, `tuple`, `set`), le découpage (*slicing*), les fonctions et les fameuses compréhensions de liste.

### 2.1 Les listes (`list`), indices négatifs et découpage (*slicing*)

* La liste est ordonnée et modifiable (équivalent de `List<T>` en C# ou `ArrayList` en Java).
* **Indice négatif** : `-1` donne le dernier élément, `-2` l'avant-dernier !
* **Découpage `liste[debut:fin:pas]`** :
  * `[:3]` : les 3 premiers éléments
  * `[1:4]` : du 2e au 4e élément (borne de fin exclue)
  * `[::-1]` : inverse toute la liste en une seule opération !

```python
# --- 2.1 Manipulation des listes et découpage (slicing) ---

ports_services = [22, 80, 443, 3306, 5432, 8080]

# Ajout d'un élément à la fin avec .append()
ports_services.append(9090)

print("Liste complète des ports :", ports_services)
print("Premier port [0]          :", ports_services[0])
print("Dernier port [-1]         :", ports_services[-1])
print("Les 3 premiers [:3]       :", ports_services[:3])
print("Du 2e au 4e port [1:4]    :", ports_services[1:4])
print("Liste inversée [::-1]     :", ports_services[::-1])
```

### 2.2 Dictionnaires (`dict`), Tuples (`tuple`) et Ensembles (`set`)

* **`dict` (Dictionnaire)** : Table de hachage clé-valeur avec recherche instantanée en $O(1)$. Très proche d'un objet JSON. Utiliser `.get()` pour éviter une erreur si la clé n'existe pas.
* **`tuple` (Tuple)** : Séquence fixe et **immuable** (non modifiable). Idéal pour des coordonnées ou pour retourner plusieurs valeurs d'une fonction.
* **`set` (Ensemble)** : Collection non ordonnée qui **élimine automatiquement tous les doublons** !

```python
# --- 2.2 Dictionnaires, Tuples et Ensembles ---

# 1. Dictionnaire (équivalent Map / Hashtable / Objet JSON)
serveur_config = {
    "nom": "srv-base-donnees",
    "ip": "192.168.10.50",
    "cpu_coeurs": 16,
    "memoire_go": 64,
    "sauvegarde_active": True,
}

# Accès direct et accès sécurisé avec .get()
nom_srv = serveur_config["nom"]
datacenter = serveur_config.get("datacenter", "Montréal (Défaut)")
print(f"Serveur {nom_srv} localisé à : {datacenter}")

# 2. Tuple (immuable) et déballage direct dans des variables (unpacking)
coord_montreal = (45.5017, -73.5673)
latitude, longitude = coord_montreal
print(f"Coordonnées GPS : Latitude={latitude}, Longitude={longitude}")

# 3. Ensemble (set) : dédoublonnage automatique instantané
adresses_ip_brutes = [
    "10.0.0.1", "10.0.0.2", "10.0.0.1", "192.168.1.15", 
    "10.0.0.2", "172.16.0.4", "10.0.0.1"
]
adresses_ip_uniques = set(adresses_ip_brutes)

print(f"IPs brutes analysées : {len(adresses_ip_brutes)} adresses")
print(f"IPs uniques filtrées : {len(adresses_ip_uniques)} adresses ->", adresses_ip_uniques)
```

### 2.3 Fonctions, valeurs par défaut et annotations de type

* Définies avec le mot-clé **`def`**.
* Acceptent des **valeurs par défaut** pour les paramètres optionnels.
* Permettent d'appeler les arguments par leur nom (**keyword arguments**).
* Peuvent retourner plusieurs valeurs d'un coup sous forme de tuple !

```python
# --- 2.3 Fonctions et retours multiples ---

def estimer_cout_infonuagique(
    nb_heures: int,
    prix_heure: float = 0.45,
    inclure_taxes: bool = True
) -> tuple:
    """Calcule le coût sous-total et total pour un serveur cloud."""
    sous_total = nb_heures * prix_heure
    taux_taxes = 0.14975 if inclure_taxes else 0.0  # TPS + TVQ au Québec
    total_avec_taxes = sous_total * (1 + taux_taxes)
    
    # Retour de deux valeurs d'un coup !
    return round(sous_total, 2), round(total_avec_taxes, 2)

# Appel avec paramètres nommés :
st, total = estimer_cout_infonuagique(nb_heures=720, prix_heure=0.50)
print(f"Facture pour 720 h : Sous-total = {st} $ | Total facturé = {total} $")

# Appel avec argument par défaut sans taxes :
st_net, total_net = estimer_cout_infonuagique(nb_heures=100, inclure_taxes=False)
print(f"Facture sans taxes : {total_net} $")
```

### 2.4 Le style « pythonique » : compréhensions de liste et bloc `with`

* **Compréhension de liste** : Remplace une boucle `for` avec un `.append()` par une expression lisible sur une seule ligne (l'équivalent de `.Where().Select()` en C# LINQ ou `.filter().map()` en JS).
* **Gestionnaire de contexte `with`** : Assure la libération automatique des ressources (comme la fermeture d'un fichier ou d'une connexion), même si une erreur survient (équivalent de `using` en C#).

```python
# --- 2.4 Compréhension de liste et gestionnaire de contexte ---

mesures_latences_ms = [12, 145, 28, 310, 42, 85, 220, 18]

# 1. Compréhension de liste : filtrer les latences anormales (> 100 ms) et les formater
alertes_latence = [f"{ms} ms" for ms in mesures_latences_ms if ms > 100]
print("Latences critiques détectées en 1 ligne :", alertes_latence)

# 2. Utilisation de 'with' pour écrire un rapport texte
nom_fichier = "journal_audit.txt"
with open(nom_fichier, "w", encoding="utf-8") as f:
    f.write("=== RAPPORT D'AUDIT SYSTÈME ===\n")
    f.write(f"Nombre de dépassements de seuil : {len(alertes_latence)}\n")
# Le fichier est déjà fermé automatiquement et proprement ici !

# Vérification du contenu écrit :
with open(nom_fichier, "r", encoding="utf-8") as f:
    print("\nContenu du fichier généré :\n" + f.read())
```

### 2.5 Exemple intégrateur 2 : Traitement d'un lot de billets de support TI

Voici un cas d'usage typique en TI : traiter une liste de dictionnaires représentant des billets d'assistance, extraire les catégories uniques, filtrer les urgences non résolues et calculer des statistiques clés.

```python
# --- 2.5 Gestionnaire de billets de support TI ---

billets_support = [
    {"id": 101, "titre": "Panne VPN télétravail", "categorie": "Réseau", "priorite": "Haute", "temps_min": 45, "resolu": False},
    {"id": 102, "titre": "Réinitialisation mot de passe AD", "categorie": "Accès", "priorite": "Basse", "temps_min": 10, "resolu": True},
    {"id": 103, "titre": "Certificat SSL expiré sur portail", "categorie": "Sécurité", "priorite": "Haute", "temps_min": 30, "resolu": True},
    {"id": 104, "titre": "Lenteur base de données ERP", "categorie": "Base de données", "priorite": "Haute", "temps_min": 95, "resolu": False},
    {"id": 105, "titre": "Ajout licence Microsoft 365", "categorie": "Accès", "priorite": "Moyenne", "temps_min": 15, "resolu": True},
    {"id": 106, "titre": "Pare-feu bloque le port 443", "categorie": "Réseau", "priorite": "Haute", "temps_min": 60, "resolu": False},
]

# 1. Extraire les catégories uniques avec un ensemble (set)
categories_uniques = {b["categorie"] for b in billets_support}
print("1. Catégories uniques rencontrées :", categories_uniques)

# 2. Compréhension de liste : filtrer les titres des urgences (Haute priorité) non résolues
urgences_ouvertes = [
    b["titre"] for b in billets_support
    if b["priorite"] == "Haute" and not b["resolu"]
]
print("\n2. Billets urgents nécessitant une intervention :")
for titre in urgences_ouvertes:
    print(f"   • {titre}")

# 3. Fonction calculant un résumé complet
def produire_statistiques_support(liste: list) -> tuple:
    nb_total = len(liste)
    nb_resolus = sum(1 for b in liste if b["resolu"])
    temps_total = sum(b["temps_min"] for b in liste)
    temps_moyen = round(temps_total / nb_total, 1)
    return nb_total, nb_resolus, temps_moyen

total, resolus, moyenne = produire_statistiques_support(billets_support)
print(f"\n3. Bilan de performance : {resolus}/{total} billets résolus | Temps moyen : {moyenne} minutes")
```

### 3.1 Sécurité TI : Générateur de mot de passe cryptographique (4 lignes)
Grâce au module standard `secrets`, conçu expressément pour la cryptographie (contrairement à `random`).

```python
# --- 3.1 Génération de mot de passe sécurisé ---

import secrets
import string

caracteres = string.ascii_letters + string.digits + "!@#$%^&*"
mot_de_passe = "".join(secrets.choice(caracteres) for _ in range(16))

print("Mot de passe cryptographiquement sécurisé (16 caractères) :")
print(mot_de_passe)
```

### 3.2 Interrogation d'une API REST JSON en direct (Météo au Québec)
En quelques lignes, sans aucune bibliothèque tierce (uniquement la bibliothèque standard `urllib` et `json`), nous interrogeons l'API publique Open-Meteo pour obtenir les températures en temps réel.

```python
# --- 3.2 Requête API REST en direct ---

import json
import urllib.request

villes_quebec = {
    "Montréal": (45.5017, -73.5673),
    "Québec": (46.8139, -71.2080),
    "Gatineau": (45.4765, -75.7013),
    "Saguenay": (48.4284, -71.0685),
}

print("🌤 Températures actuelles en direct (API Open-Meteo) :")
for ville, (lat, lon) in villes_quebec.items():
    url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&current_weather=true"
    try:
        with urllib.request.urlopen(url, timeout=5) as reponse:
            donnees = json.loads(reponse.read().decode("utf-8"))
            meteo = donnees["current_weather"]
            temp = meteo["temperature"]
            vent = meteo["windspeed"]
            print(f"  • {ville:<10} : {temp:>4.1f} °C (Vent : {vent} km/h)")
    except Exception as err:
        print(f"  • {ville:<10} : Simulation hors-ligne (12.0 °C)")
```

### 3.3 Science des données et visualisation graphique instantanée avec `pandas`
Google Colab inclut déjà `pandas` et `matplotlib`. Regardez comment transformer notre lot de billets TI en tableau interactif, calculer des agrégats par catégorie et tracer un graphique professionnel :

```python
# --- 3.3 Analyse de données et graphique Pandas ---

import pandas as pd
import matplotlib.pyplot as plt

# 1. Conversion de notre liste de dictionnaires en DataFrame Pandas
df_billets = pd.DataFrame(billets_support)

# 2. Agrégation en une ligne : temps total et moyen par catégorie
rapport_categories = df_billets.groupby("categorie")["temps_min"].agg(
    nb_billets="count",
    temps_total_min="sum",
    temps_moyen_min="mean"
).reset_index()

print("Tableau d'analyse par catégorie de billet TI :")
print(rapport_categories.to_string(index=False))

# 3. Graphique horizontal généré directement
plt.figure(figsize=(7, 3.5))
plt.barh(rapport_categories["categorie"], rapport_categories["temps_total_min"], color="#0284c7")
plt.title("Temps total consacré par catégorie de billet (minutes)")
plt.xlabel("Minutes")
plt.ylabel("Catégorie")
plt.grid(axis="x", linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
